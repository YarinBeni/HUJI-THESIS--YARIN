## 4. Key Metrics: Texts vs Words

**Important distinction:**
- Each **row** = 1 word
- Each **fragment_id** = 1 text/tablet
- We need to count both to understand the data

```python
def get_corpus_stats(df, name):
    """Calculate key statistics for a corpus dataframe."""
    n_words = len(df)
    n_texts = df['fragment_id'].nunique()
    n_domains = df['domain'].nunique()
    n_places = df['place_discovery'].nunique()
    avg_words_per_text = n_words / n_texts if n_texts > 0 else 0
    
    return {
        'Corpus': name,
        'Total Words': f"{n_words:,}",
        'Unique Texts': f"{n_texts:,}",
        'Avg Words/Text': f"{avg_words_per_text:.1f}",
        'Unique Domains': n_domains,
        'Unique Places': n_places
    }

stats = [
    get_corpus_stats(archibab, 'archibab (OB/Group 1)'),
    get_corpus_stats(oracc, 'oracc (NA/Group 2)'),
    get_corpus_stats(lbl, 'lbl (LB/Group 3)')
]

# Calculate totals across all corpora
total_words = len(archibab) + len(oracc) + len(lbl)
total_texts = archibab['fragment_id'].nunique() + oracc['fragment_id'].nunique() + lbl['fragment_id'].nunique()
total_avg = total_words / total_texts if total_texts > 0 else 0

# For domains and places, we need to count unique across all corpora
all_combined = pd.concat([archibab, oracc, lbl], ignore_index=True)
total_domains = all_combined['domain'].nunique()
total_places = all_combined['place_discovery'].nunique()

# Add totals row
stats.append({
    'Corpus': 'TOTAL (All Groups)',
    'Total Words': f"{total_words:,}",
    'Unique Texts': f"{total_texts:,}",
    'Avg Words/Text': f"{total_avg:.1f}",
    'Unique Domains': total_domains,
    'Unique Places': total_places
})

print("CORPUS STATISTICS SUMMARY")
print("=" * 80)
stats_df = pd.DataFrame(stats)
display(stats_df)
```

## 5. Explore `domain` Column (Genre)

**Why this matters**: We decided to focus on **letters only** to avoid genre-based confounders.

**Enhanced in v4**: Shows both word counts AND fragment counts with percentages.

```python
def analyze_domain(df, name):
    """Analyze domain distribution with both word and fragment counts."""
    print(f"DOMAIN (GENRE) ANALYSIS: {name}")
    print("=" * 80)
    
    # Word-level stats
    domain_words = df['domain'].value_counts()
    domain_words_pct = (df['domain'].value_counts(normalize=True) * 100).round(2)
    
    # Fragment-level stats (distinct texts)
    domain_fragments = df.groupby('domain')['fragment_id'].nunique()
    total_fragments = df['fragment_id'].nunique()
    domain_fragments_pct = (domain_fragments / total_fragments * 100).round(2)
    
    # Combine into single dataframe
    domain_df = pd.DataFrame({
        'Words': domain_words,
        'Words %': domain_words_pct,
        'Fragments': domain_fragments,
        'Fragments %': domain_fragments_pct
    }).sort_values('Words', ascending=False)
    
    # Format percentages for display
    domain_df_display = domain_df.copy()
    domain_df_display['Words %'] = domain_df_display['Words %'].astype(str) + '%'
    domain_df_display['Fragments %'] = domain_df_display['Fragments %'].astype(str) + '%'
    
    print(f"\nDomain distribution (words AND fragments):")
    print(f"Total words: {len(df):,} | Total fragments: {total_fragments:,}")
    display(domain_df_display)
    
    # DATA-DRIVEN CONCLUSION
    n_unique_domains = df['domain'].nunique()
    domain_values = df['domain'].unique()
    
    print("\n" + "-" * 80)
    print(f"DATA-DRIVEN CONCLUSION:")
    print(f"  - Number of unique domain values: {n_unique_domains}")
    print(f"  - Domain values: {list(domain_values)}")
    
    if n_unique_domains == 1:
        print(f"  ✓ Only 1 domain value exists ('{domain_values[0]}')")
        print(f"  ✓ No filtering needed - filtering would return 100% of data")
    else:
        # Check if all domains contain 'letter' or 'lettre' (case insensitive)
        letter_domains = [d for d in domain_values if d and ('letter' in str(d).lower() or 'lettre' in str(d).lower())]
        non_letter_domains = [d for d in domain_values if d and d not in letter_domains]
        
        print(f"  - Domains containing 'letter'/'lettre': {len(letter_domains)} → {letter_domains}")
        print(f"  - Other domains: {len(non_letter_domains)} → {non_letter_domains}")
        
        if len(non_letter_domains) > 0:
            non_letter_word_count = domain_df.loc[domain_df.index.isin(non_letter_domains), 'Words'].sum()
            non_letter_pct = non_letter_word_count / len(df) * 100
            print(f"  ⚠️ Non-letter domains contain {non_letter_word_count:,} words ({non_letter_pct:.1f}%)")
            print(f"  ⚠️ FILTERING NEEDED to get letters only")
        else:
            print(f"  ✓ All {n_unique_domains} domains appear to be letter types")
            print(f"  ✓ No filtering needed for letters-only analysis")
    
    return domain_df
```

## 5.1 Investigating Small Domain Categories (archibab)

Some domain values have very few samples. Let's investigate if they should be merged or excluded.

```python
print("INVESTIGATING SMALL DOMAIN CATEGORIES (archibab)")
print("=" * 80)

# Find small categories (threshold: < 1% of total words OR < 100 words)
WORD_THRESHOLD = 100
PCT_THRESHOLD = 1.0

total_words = len(archibab)
total_fragments = archibab['fragment_id'].nunique()

small_domains = archibab.groupby('domain').agg({
    'value': 'count',
    'fragment_id': 'nunique'
}).rename(columns={'value': 'word_count', 'fragment_id': 'fragment_count'})

small_domains['word_pct'] = (small_domains['word_count'] / total_words * 100).round(2)
small_domains['fragment_pct'] = (small_domains['fragment_count'] / total_fragments * 100).round(2)

# Filter to small categories
small_domains = small_domains[
    (small_domains['word_count'] < WORD_THRESHOLD) | 
    (small_domains['word_pct'] < PCT_THRESHOLD)
].sort_values('word_count')

print(f"\nDomains with < {WORD_THRESHOLD} words OR < {PCT_THRESHOLD}% of total:")
print(f"(Total corpus: {total_words:,} words, {total_fragments:,} fragments)")
display(small_domains)
```

```python
print("\n" + "=" * 80)
print("DETAILED EXAMINATION OF EACH SMALL CATEGORY")
print("=" * 80)

for domain_name in small_domains.index:
    subset = archibab[archibab['domain'] == domain_name]
    word_count = len(subset)
    fragment_count = subset['fragment_id'].nunique()
    fragment_ids = subset['fragment_id'].unique().tolist()
    places = subset['place_discovery'].unique().tolist()
    
    print(f"\n{'─' * 70}")
    print(f"Domain: '{domain_name}'")
    print(f"{'─' * 70}")
    print(f"  Words: {word_count:,}")
    print(f"  Fragments: {fragment_count}")
    print(f"  Fragment IDs: {fragment_ids[:10]}{'...' if len(fragment_ids) > 10 else ''}")
    print(f"  Places: {places[:5]}{'...' if len(places) > 5 else ''}")
    
    # Show sample words
    print(f"  Sample words:")
    display(subset[['fragment_id', 'fragment_line_num', 'value', 'clean_value', 'lemma']].head(5))
    
    # DATA-DRIVEN RECOMMENDATION
    domain_lower = str(domain_name).lower()
    
    # Check for typo variants
    if 'administratif' in domain_lower:
        print(f"  → RECOMMENDATION: Likely typo of 'lettre administrative'. Merge into 'lettre administrative'.")
    elif 'personnelle' in domain_lower or '(personnelle)' in domain_lower:
        print(f"  → RECOMMENDATION: Variant of 'lettre privée'. Merge into 'lettre privée'.")
    elif 'unknown' in domain_lower or 'inconnu' in domain_lower:
        print(f"  → RECOMMENDATION: Unknown category. Review manually or exclude from analysis.")
    elif word_count < 50:
        print(f"  → RECOMMENDATION: Very small category ({word_count} words). Consider merging or excluding.")
    else:
        print(f"  → RECOMMENDATION: Review manually to determine appropriate action.")
```

```python
print("\n" + "=" * 80)
print("SUMMARY: RECOMMENDED DOMAIN MERGES FOR ARCHIBAB")
print("=" * 80)

# Build recommendations based on data
all_domains = archibab['domain'].unique()

merge_recommendations = []

for d in all_domains:
    d_lower = str(d).lower() if d else ''
    word_count = len(archibab[archibab['domain'] == d])
    
    if 'administratif' in d_lower and 'administrative' not in d_lower:
        merge_recommendations.append({
            'original': d,
            'merge_into': 'lettre administrative',
            'reason': 'typo variant',
            'word_count': word_count
        })
    elif 'privée' in d_lower and 'personnelle' in d_lower:
        merge_recommendations.append({
            'original': d,
            'merge_into': 'lettre privée',
            'reason': 'more specific variant',
            'word_count': word_count
        })
    elif 'unknown' in d_lower or 'inconnu' in d_lower:
        merge_recommendations.append({
            'original': d,
            'merge_into': 'Unknown',
            'reason': 'standardize unknown labels',
            'word_count': word_count
        })

if merge_recommendations:
    merge_df = pd.DataFrame(merge_recommendations)
    display(merge_df)
    
    total_affected = merge_df['word_count'].sum()
    print(f"\nTotal words affected by merges: {total_affected:,} ({total_affected/len(archibab)*100:.2f}% of corpus)")
else:
    print("No merge recommendations identified.")
```

## 5.3 Applying Domain Merges for Archibab

Now we'll apply the recommended merges to standardize the domain labels in archibab.

```python
print("APPLYING DOMAIN MERGES TO ARCHIBAB")
print("=" * 80)

# Create a copy for merging
archibab_merged = archibab.copy()

# Define the merges based on the recommendations
domain_merge_map = {
    'inconnu, lettre cassée': 'Unknown',
    'lettre administratif': 'lettre administrative',
    'UNKNOWN': 'Unknown',
    'lettre privée (personnelle)': 'lettre privée'
}

# Show before state
print("\nBEFORE merges:")
before_domains = archibab_merged['domain'].value_counts().to_dict()
print(f"  Number of unique domains: {archibab_merged['domain'].nunique()}")

# Apply the merges
for original, target in domain_merge_map.items():
    n_words = len(archibab_merged[archibab_merged['domain'] == original])
    if n_words > 0:
        archibab_merged.loc[archibab_merged['domain'] == original, 'domain'] = target
        print(f"\n  ✓ Merged '{original}' → '{target}': {n_words:,} words")

# Show after state
print("\n" + "-" * 80)
print("AFTER merges:")
after_domains = archibab_merged['domain'].value_counts()
print(f"  Number of unique domains: {archibab_merged['domain'].nunique()}")
print("\n  Updated domain distribution:")

# Create full comparison table
domain_words = after_domains
domain_words_pct = (domain_words / len(archibab_merged) * 100).round(2)
domain_fragments = archibab_merged.groupby('domain')['fragment_id'].nunique()
total_fragments = archibab_merged['fragment_id'].nunique()
domain_fragments_pct = (domain_fragments / total_fragments * 100).round(2)

merged_domain_df = pd.DataFrame({
    'Words': domain_words,
    'Words %': domain_words_pct.astype(str) + '%',
    'Fragments': domain_fragments.reindex(domain_words.index),
    'Fragments %': domain_fragments_pct.reindex(domain_words.index).astype(str) + '%'
}).sort_values('Words', ascending=False)

display(merged_domain_df)

# Verification
print("\n" + "=" * 80)
print("VERIFICATION:")
domains_before = len(before_domains)
domains_after = archibab_merged['domain'].nunique()
domains_reduced = domains_before - domains_after

if domains_reduced == len(domain_merge_map):
    print(f"  ✓ Successfully reduced {domains_reduced} domain categories (from {domains_before} to {domains_after})")
    print(f"  ✓ All {len(domain_merge_map)} recommended merges applied")
else:
    print(f"  ⚠️ Expected to reduce by {len(domain_merge_map)} but reduced by {domains_reduced}")
    
# Replace archibab with the merged version for subsequent cells
archibab = archibab_merged
print(f"\n✓ archibab updated with merged domains")
```

## 5.2 Domain Standardization Plan

**Problem**: Domain values are inconsistent across corpora:
- archibab: Fine-grained (lettre administrative, politique, privée, diplomatique)
- oracc: Coarse (NALet, NAAdm)
- lbl: Single value (LB Letters)

**Solution**: Create two columns:
1. `domain_original` - Keep fine-grained data
2. `domain_standard` - Unified for cross-corpus comparison

```python
print("DOMAIN STANDARDIZATION PLAN")
print("=" * 80)

# Collect all unique domains across all corpora
all_archibab_domains = archibab['domain'].unique().tolist()
all_oracc_domains = oracc['domain'].unique().tolist()
all_lbl_domains = lbl['domain'].unique().tolist()

print(f"\nUnique domains in each corpus:")
print(f"  archibab: {len(all_archibab_domains)} values → {all_archibab_domains}")
print(f"  oracc: {len(all_oracc_domains)} values → {all_oracc_domains}")
print(f"  lbl: {len(all_lbl_domains)} values → {all_lbl_domains}")
```

```python
# Define the standardization mapping
domain_to_standard = {}
domain_to_finegrained = {}

# Build mapping dynamically based on actual data
for d in all_archibab_domains:
    d_str = str(d) if d else 'Unknown'
    d_lower = d_str.lower()
    
    # Standard mapping (coarse)
    if 'lettre' in d_lower or 'letter' in d_lower:
        domain_to_standard[d] = 'Letter'
    elif 'unknown' in d_lower or 'inconnu' in d_lower:
        domain_to_standard[d] = 'Unknown'
    else:
        domain_to_standard[d] = 'Other'
    
    # Fine-grained mapping (with typo fixes)
    if 'administratif' in d_lower and 'administrative' not in d_lower:
        domain_to_finegrained[d] = 'Administrative Letter'
    elif 'administrative' in d_lower:
        domain_to_finegrained[d] = 'Administrative Letter'
    elif 'politique' in d_lower:
        domain_to_finegrained[d] = 'Political Letter'
    elif 'privée' in d_lower and 'personnelle' in d_lower:
        domain_to_finegrained[d] = 'Private Letter'
    elif 'privée' in d_lower:
        domain_to_finegrained[d] = 'Private Letter'
    elif 'diplomatique' in d_lower:
        domain_to_finegrained[d] = 'Diplomatic Letter'
    elif 'unknown' in d_lower or 'inconnu' in d_lower:
        domain_to_finegrained[d] = 'Unknown'
    else:
        domain_to_finegrained[d] = d_str

# Add oracc domains
for d in all_oracc_domains:
    d_str = str(d) if d else 'Unknown'
    if 'NALet' in d_str:
        domain_to_standard[d] = 'Letter'
        domain_to_finegrained[d] = 'Neo-Assyrian Letter'
    elif 'NAAdm' in d_str:
        domain_to_standard[d] = 'Administrative'
        domain_to_finegrained[d] = 'Neo-Assyrian Administrative'
    else:
        domain_to_standard[d] = 'Other'
        domain_to_finegrained[d] = d_str

# Add lbl domains
for d in all_lbl_domains:
    d_str = str(d) if d else 'Unknown'
    if 'letter' in d_str.lower():
        domain_to_standard[d] = 'Letter'
        domain_to_finegrained[d] = 'Late Babylonian Letter'
    else:
        domain_to_standard[d] = 'Other'
        domain_to_finegrained[d] = d_str

print("\nSTANDARD MAPPING (domain → domain_standard):")
print("-" * 60)
for k, v in sorted(domain_to_standard.items(), key=lambda x: str(x[0])):
    print(f"  '{k}' → '{v}'")

print("\nFINE-GRAINED MAPPING (domain → domain_finegrained):")
print("-" * 60)
for k, v in sorted(domain_to_finegrained.items(), key=lambda x: str(x[0])):
    print(f"  '{k}' → '{v}'")
```

```python
# Verify the mapping coverage
print("\nVERIFYING MAPPING COVERAGE")
print("=" * 80)

for corpus_name, df in [('archibab', archibab), ('oracc', oracc), ('lbl', lbl)]:
    unmapped = [d for d in df['domain'].unique() if d not in domain_to_standard]
    mapped_count = len([d for d in df['domain'].unique() if d in domain_to_standard])
    total_count = df['domain'].nunique()
    
    if len(unmapped) == 0:
        print(f"  ✓ {corpus_name}: All {total_count} domains mapped")
    else:
        print(f"  ⚠️ {corpus_name}: {len(unmapped)} unmapped domains: {unmapped}")
```

## 6. Explore `place_discovery` Column (Provenance)

**Why this matters**: Nathan identified that provenance determines temporal period:

| Locations | Period | Group |
|-----------|--------|-------|
| Mari, Larsa, Sippar, Ur | 2nd millennium BCE | Group 1 |
| Nineveh, Assur, Calah (Nimrud) | 9-7 cent. BCE | Group 2 |
| Babylon, Borsipa, Uruk | 6-4 cent. BCE | Group 3 |

**Enhanced in v4**: Shows both word counts AND fragment counts with percentages.

```python
def analyze_provenance(df, name, expected_locations):
    """Analyze provenance distribution with both word and fragment counts."""
    print(f"PROVENANCE ANALYSIS: {name}")
    print("=" * 80)
    
    # Word-level stats
    place_words = df['place_discovery'].value_counts()
    place_words_pct = (df['place_discovery'].value_counts(normalize=True) * 100).round(2)
    
    # Fragment-level stats (distinct texts)
    place_fragments = df.groupby('place_discovery')['fragment_id'].nunique()
    total_fragments = df['fragment_id'].nunique()
    place_fragments_pct = (place_fragments / total_fragments * 100).round(2)
    
    # Combine into single dataframe
    place_df = pd.DataFrame({
        'Words': place_words,
        'Words %': place_words_pct,
        'Fragments': place_fragments,
        'Fragments %': place_fragments_pct
    }).sort_values('Words', ascending=False)
    
    # Format percentages for display
    place_df_display = place_df.copy()
    place_df_display['Words %'] = place_df_display['Words %'].astype(str) + '%'
    place_df_display['Fragments %'] = place_df_display['Fragments %'].astype(str) + '%'
    
    print(f"\nProvenance distribution (words AND fragments):")
    print(f"Total words: {len(df):,} | Total fragments: {total_fragments:,}")
    display(place_df_display)
    
    # DATA-DRIVEN VERIFICATION against expected locations
    print("\n" + "-" * 80)
    print(f"VERIFICATION against expected locations: {expected_locations}")
    print("-" * 80)
    
    found_locations = []
    not_found_locations = []
    
    for loc in expected_locations:
        matches = place_df[place_df.index.str.contains(loc, case=False, na=False)]
        if len(matches) > 0:
            found_locations.append(loc)
            print(f"  ✓ {loc}: FOUND")
            for idx, row in matches.iterrows():
                print(f"      → '{idx}': {row['Words']:,} words, {row['Fragments']} fragments")
        else:
            not_found_locations.append(loc)
            print(f"  ✗ {loc}: NOT FOUND")
    
    # DATA-DRIVEN CONCLUSION
    print("\n" + "-" * 80)
    print("DATA-DRIVEN CONCLUSION:")
    found_ratio = len(found_locations) / len(expected_locations) * 100
    print(f"  - Expected locations found: {len(found_locations)}/{len(expected_locations)} ({found_ratio:.0f}%)")
    
    if found_ratio >= 50:
        print(f"  ✓ Majority of expected locations present - corpus aligns with expected group")
    else:
        print(f"  ⚠️ Less than 50% of expected locations found - verify corpus assignment")
    
    if not_found_locations:
        print(f"  - Missing locations: {not_found_locations}")
    
    return place_df
```

```python
print("CHECKING FOR INDEX_IN_LINE GAPS (Broken Words) - ALL CORPORA")
print("="*80)

def check_index_gaps(df, name):
    """Check for gaps in index_in_line within each line."""
    gaps = []
    
    for (fid, line_num), group in df.groupby(['fragment_id', 'fragment_line_num']):
        indices = sorted(group['index_in_line'].values)
        for i in range(len(indices) - 1):
            if indices[i+1] - indices[i] > 1:
                gaps.append({
                    'fragment_id': fid,
                    'line_num': line_num,
                    'gap_from': indices[i],
                    'gap_to': indices[i+1],
                    'missing_count': indices[i+1] - indices[i] - 1
                })
    
    gaps_df = pd.DataFrame(gaps)
    return gaps_df

def analyze_gaps(df, name):
    """Analyze and report gaps for a corpus."""
    print(f"\n{'─'*70}")
    print(f"📁 {name}")
    print(f"{'─'*70}")
    
    gaps_df = check_index_gaps(df, name)
    
    total_lines = df.groupby(['fragment_id', 'fragment_line_num']).ngroups
    total_texts = df['fragment_id'].nunique()
    total_words = len(df)
    
    print(f"  Total texts: {total_texts:,}")
    print(f"  Total lines: {total_lines:,}")
    print(f"  Total words: {total_words:,}")
    
    if len(gaps_df) > 0:
        texts_with_gaps = gaps_df['fragment_id'].nunique()
        lines_with_gaps = len(gaps_df)
        total_missing_words = gaps_df['missing_count'].sum()
        
        print(f"\n  Gap statistics:")
        print(f"    - Lines with gaps: {lines_with_gaps:,} ({lines_with_gaps/total_lines*100:.2f}% of lines)")
        print(f"    - Texts with at least one gap: {texts_with_gaps:,} ({texts_with_gaps/total_texts*100:.1f}% of texts)")
        print(f"    - Total estimated missing words: {total_missing_words:,}")
        print(f"    - Avg missing words per gap: {total_missing_words/lines_with_gaps:.2f}")
        
        print(f"\n  Gap size distribution:")
        gap_dist = gaps_df['missing_count'].value_counts().sort_index()
        for size, count in gap_dist.head(10).items():
            print(f"    Gap of {size} word(s): {count:,} occurrences")
        if len(gap_dist) > 10:
            print(f"    ... and {len(gap_dist) - 10} more gap sizes")
        
        print(f"\n  Sample gaps:")
        display(gaps_df.head(10))
    else:
        print(f"\n  ✓ No gaps found - all indices are consecutive")
    
    return gaps_df

# Reload data for analysis
print("Analyzing full corpora (this may take a moment)...")
archibab = pd.read_csv(DATA_DIR / 'archibab_nor.csv')
oracc = pd.read_csv(DATA_DIR / 'oracc_let_adm_nor.csv')
lbl = pd.read_csv(DATA_DIR / 'lbl_nor.csv')
```
